# Personal Budget Tracker

- ***Description:*** This is a personal budget tracker for ***FinTrack***; it will collect monthly income and expenses 
 and produce a formatted summary.

## Section 1: Analyst information
- Enter your name and the month being analysed when prompted.

```python
analyst_name = input("Enter Analyst name:     ")
budget_month = input("Enter the Budget month: ")
```

## Section 2: Monthly income and expenses
- Enter the income and monthly expenses when prompted.

```python
monthly_income = float(input("Enter Monthly income $: "))
rent = float(input("Enter Rent Expense $: "))
food = float(input("Enter Food & groceries Expenses $: "))
transport = float(input("Enter Transport Expense $: "))
utilities = float(input("Enter Utilities Expenses $: "))
```

### Section 3: Budget calculations
- Total expenses, remaining budget, and savings rate are computed
using plain Python arithmetic.

```python
total_expenses = rent + food + transport + utilities
remaining_budget = total_expenses - monthly_income
saving_rate = (remaining_budget / monthly_income) * 100
```

#### Section 4: Budget summary report
- The report below is generated using Python f-strings.

```python
print("====== FinTrack Budget Report ======")
print(f"Analyst Name              : {analyst_name}")
print(f"Budget Month              : {budget_month}")
print("------------------------------------")
print(f"Monthly Income            : ${monthly_income:,.2f}")
print(f"Rent Expenses             : ${rent:,.2f}")
print(f"Food & groceries Expenses : ${food:,.2f}")
print(f"Transport Expenses        : ${transport:,.2f}")
print(f"Utilities Expenses        : ${utilities:,.2f}")
print("------------------------------------")
print(f"Total expenses            : ${total_expenses:,.2f}")
print(f"Remaining                 : ${remaining_budget:,.2f}")
print(f"Savings rate              : {saving_rate:.1f}%")
print("====================================")
```
